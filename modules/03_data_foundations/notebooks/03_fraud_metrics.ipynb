## Confusion-matrix interpretation

For the positive fraud class, a **true positive** catches fraud, a **false positive** blocks a legitimate payment, a **true negative** lets a legitimate payment through, and a **false negative** misses fraud. Costs are asymmetric, so accuracy alone is not a decision rule.

```python
labels = np.zeros(1_000, dtype=np.int8)
labels[-1] = 1
all_legitimate = np.zeros_like(labels)
accuracy = float((labels == all_legitimate).mean())
fraud_recall = 0.0
{"accuracy": accuracy, "accuracy_percent": f"{accuracy:.1%}", "fraud_recall": fraud_recall}
```

## Ranking metrics

ROC-AUC measures positive-versus-negative ranking across thresholds. PR-AUC emphasizes precision and recall for the rare positive class, so it is often more revealing for fraud. Tied scores must be consumed as one threshold group.

```python
scores = np.array([0.05, 0.10, 0.40, 0.40, 0.92])
thresholds = np.unique(scores)[::-1]
thresholds
```
