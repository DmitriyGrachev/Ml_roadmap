## Array model

The completed example models branch operations rather than the earlier matrix exercise. Rows are branches; columns are queue minutes, retry rate, and normalized load. Vectorization applies one operation to the complete array.

```python
branch_features = np.array(
    [[8.0, 0.02, 0.70], [21.0, 0.08, 0.95], [13.0, 0.03, 0.80]],
    dtype=np.float64,
)
slow_branches = branch_features[:, 0] > 12
{
    "shape": branch_features.shape,
    "dtype": branch_features.dtype,
    "masked_rows": branch_features[slow_branches],
}
```

```python
column_scales = np.array([1.0 / 30.0, 10.0, 1.0])
scaled = branch_features * column_scales  # broadcasting across rows
daily = branch_features.reshape(1, 3, 3)
column_means = branch_features.mean(axis=0)
scaled, daily.shape, column_means
```

```python
risk_weights = np.array([0.45, 0.35, 0.20])
branch_risk_scores = scaled.dot(risk_weights)
branch_risk_scores
```
